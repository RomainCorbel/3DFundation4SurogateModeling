```python

dataset1 = torch.load('save_dataset/Global_pointnet/train_dataset', map_location="cpu", weights_only=False)
print(dataset1)

all_g1 = torch.stack([data.g for data in dataset1], dim=0)  # [N, 1024]

# mean and std per dimension
g_mean_dim1 = all_g1.float().mean(dim=0)   # [1024]
g_std_dim1  = all_g1.float().std(dim=0)    # [1024]

print("Mean per dimension shape:", g_mean_dim1.shape)
print("Std  per dimension shape:", g_std_dim1.shape)

print("mean Mean:", g_mean_dim1.mean())
print("mean Std:", g_std_dim1.mean())


print("First 20 mean values:", g_mean_dim1[:20])
print("First 20 std values:", g_std_dim1[:20])

print("Loading normalization file...")
coef_norm = torch.load(
    'save_dataset/Global_pointnet/normalization',
    map_location='cpu',
    weights_only=False
)

names = ["mean_x", "std_x", "mean_y", "std_y", "mean_g", "std_g"]

print("\nNormalization summary:")
for name, arr in zip(names, coef_norm):
    if arr is None:
        print(f"{name}: None")
    else:
        try:
            print(f"{name}: shape = {arr.shape}")
        except AttributeError:
            print(f"{name}: type = {type(arr)} (no shape attribute)")

dataset1 = torch.load(
    'save_dataset/Global_pointnet/test_dataset',
    map_location="cpu",
    weights_only=False
)
print(dataset1)

# ------------------------------------------------------
# 1) Empilement des g en [N, 1024]
# ------------------------------------------------------
all_g1 = torch.stack([data.g for data in dataset1], dim=0).float()  # [N, 1024]

# mean and std per dimension
g_mean_dim1 = all_g1.mean(dim=0)   # [1024]
g_std_dim1  = all_g1.std(dim=0)    # [1024]

print("Mean per dimension shape:", g_mean_dim1.shape)
print("Std  per dimension shape:", g_std_dim1.shape)

print("mean Mean:", g_mean_dim1.mean())
print("mean Std:", g_std_dim1.mean())

print("First 20 mean values:", g_mean_dim1[:20])
print("First 20 std values:", g_std_dim1[:20])



dataset1 = torch.load('save_dataset/Global_pointnet/val_dataset', map_location="cpu", weights_only=False)
print(dataset1)

all_g1 = torch.stack([data.g for data in dataset1], dim=0)  # [N, 1024]

# mean and std per dimension
g_mean_dim1 = all_g1.float().mean(dim=0)   # [1024]
g_std_dim1  = all_g1.float().std(dim=0)    # [1024]

print("Mean per dimension shape:", g_mean_dim1.shape)
print("Std  per dimension shape:", g_std_dim1.shape)

print("mean Mean:", g_mean_dim1.mean())
print("mean Std:", g_std_dim1.mean())


print("First 20 mean values:", g_mean_dim1[:20])
print("First 20 std values:", g_std_dim1[:20])

```
