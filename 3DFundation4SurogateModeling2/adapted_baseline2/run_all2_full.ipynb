```python
import json
import os

# Base directory where your metric folders are stored
base_dir = "."

# Loop over the 5 MLP result folders
for i in range(1, 6):
    json_path = os.path.join(base_dir, f"metrics_baselineMLP_full{i}", "full", "MLP", "MLP_log.json")
    
    try:
        with open(json_path, "r") as f:
            data = json.load(f)
        print(f"\n=== MLP{i} Results ===")
        print(json.dumps(data, indent=4))
    except FileNotFoundError:
        print(f"\n[Warning] File not found: {json_path}")
    except json.JSONDecodeError:
        print(f"\n[Error] Could not decode JSON in: {json_path}")
```

```python
# Reference (paper): 1.13 ± 0.14 ×10⁻¹

import json, os, numpy as np, matplotlib.pyplot as plt

train, val = [], []
for i in range(1, 6):
    path = os.path.join(f"metrics_baselineMLP_full{i}", "full", "MLP", "MLP_log.json")
    with open(path) as f:
        data = json.load(f)
    train.append(data["train_loss"])
    val.append(data["val_loss"])

train, val = np.array(train), np.array(val)

mean_train, std_train = train.mean(), train.std(ddof=1)
mean_val, std_val = val.mean(), val.std(ddof=1)
cv_train = std_train / mean_train * 100
cv_val = std_val / mean_val * 100

print("=== Pressure Surface MSE Stability ===")
for i in range(5):
    print(f"MLP{i+1}: train={train[i]:.3e}, val={val[i]:.3e}")

print(f"\nTrain (scarce) mean={mean_train:.3e}, std={std_train:.3e}, Coef of variation={cv_train:.2f}%")
print(f"Val(scarce)   mean={mean_val:.3e}, std={std_val:.3e}, Coef of variation={cv_val:.2f}%")
print("\nReference (paper) for Full training set: 1.13 ± 0.14 ×10⁻¹  → CV ≈ 12.4%")

x = np.arange(1, 6)
plt.bar(x - 0.2, train, width=0.4, label="Train MSE")
plt.bar(x + 0.2, val, width=0.4, label="Val MSE")
plt.xlabel("MLP Run")
plt.ylabel("MSE")
plt.title("Pressure-Surface MSE per Run")
plt.legend()
plt.tight_layout()
plt.show()
```
